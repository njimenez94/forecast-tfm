```python
import kaggle

kaggle.api.authenticate()

busquedas = ["m5 forecasting", "titanic", "house prices", "iris"]

for termino in busquedas:
    print(f"\n=== Resultados para: {termino} ===")
    datasets = kaggle.api.dataset_list(search=termino, sort_by="votes")
    for d in datasets[:5]:
        print(f"{d.title} | votos: {d.vote_count} | descargas: {d.download_count} | usabilidad: {d.usability_rating} | kernels: {d.kernel_count}")
```
